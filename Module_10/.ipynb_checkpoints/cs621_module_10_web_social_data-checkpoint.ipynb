```python
# Load the needed modules for internet connect. 
from urllib.request import urlopen, Request
```

```python
# Most web servers do not allow web scraping anymore. 
iupui_data_science = 'https://luddy.indianapolis.iu.edu/degrees/masters/data-science/index.html'
req = Request(iupui_data_science, headers={'User-Agent': 'Mozilla/5.0'})
```

```python
# Open the desired url.
response = urlopen(req)
```

```python
#Load the html contents to the local machine. 
html = response.read()
```

```python
# Close the connection.
response.close()
```

```python
# Load BeutifulSoup module.
from bs4 import BeautifulSoup
```

```python
# Parse the webpage
soup = BeautifulSoup(html, "html.parser")
```

```python
# Read the webpage contents locally.
# print(soup.prettify())
```

```python
# Store all paragraphs in a list.
contents = []
for p in soup.find_all('p'):
    contents.append(p)
```

```python
# Check the first paragraph
contents[0]
```

```python
# Checl the second paragraph
print(contents[1])
```

```python
# Display evevything in contents.
for content in contents:
    print(content.text)
```

```python
text_contents = []
for content in contents:
    if len(content.text) != 0:
        text_contents.append(content.text)
```

```python
text_contents
```

```python
# Import the json library.
import json
```

```python
# Open a json file and load its contents to a dict.
with open("Module_10/one_tweet.json", "r") as json_file:
    dict_2 = json.load(json_file)
```

```python
# how many key:value pairs?
print(len(dict_2))
```

```python
# Show the keys in dict_2.
print(dict_2.keys())
```

```python
# Show the values associated with keys.
dict_2.values()
```

```python
# Show one key-value pair.
print(list(dict_2.keys())[0], list(dict_2.values())[0])
```

```python
# Show one key-value pair of a given key.
dict_2['entities']
```

```python
# Use a for loop to traverse the dictionary. 
for k in dict_2.keys():
    print(k, dict_2[k])
```
