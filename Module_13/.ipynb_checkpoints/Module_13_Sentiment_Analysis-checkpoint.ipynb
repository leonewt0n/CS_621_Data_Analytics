```python
#Putting Reviews in a List and writing to file

import string
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

# It is much faster to initialize the lemmatizer once outside the loop
lemmatizer = WordNetLemmatizer()


amazon_reviews = []

# Open the input file for reading and the output file for writing
with open("amazon_reviews", "r", encoding="utf-8") as in_file, \
     open("processed_reviews.txt", "w", encoding="utf-8") as out_file:
    
    for line in in_file:
        # 1. Remove punctuation
        clean_line = line.translate(str.maketrans('', '', string.punctuation))
        
        # 2. Tokenize the line into actual words
        words = word_tokenize(clean_line)
        
        processed_words = []
        for word in words:
            # 3. Lowercase the word
            word = word.lower()
            
            # 4. Lemmatize (you must assign the result back to a variable)
            #word = lemmatizer.lemmatize(word)
            
            processed_words.append(word)
            
        # Add the list of processed words to your array
        amazon_reviews.append(processed_words)
        
        # 5. Write the processed line to the text file
        # Join the words back together with a space, and add a newline at the end
        out_file.write(" ".join(processed_words) + "\n")

print("Processing complete. Saved to processed_reviews.txt")
```

```python
#Find Positive reviews and negative reviews

posi_review = 0
neg_review = 0
for review in amazon_reviews:
    positive_score  = 0
    negative_score = 0
    for w in review:
        w = w.lower()
        if w in pos_dict:
            positive_score = positive_score + 1
        if w in neg_dict:
            negative_score = negative_score + 1
    if positive_score > negative_score:
        posi_review +=1
    else:
        neg_review +=1
print(f"Positive Score: {posi_review}")
print(f"Negative Score: {neg_review}")
```

```python
import string

# 1. Load the positive and negative word lists into sets for fast lookup

with open('positive_words', 'r', encoding='utf-8') as f:
    pos_words = set(f.read().splitlines())

with open('negative_words', 'r', encoding='utf-8') as f:
    neg_words = set(f.read().splitlines())

pos_reviews = 0
neg_reviews = 0
neu_reviews = 0

# 2. Read the reviews file line by line
with open('amazon_reviews', 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue  # Skip empty lines
            
        # Lowercase and remove punctuation (adjust if your course uses simple .split() instead)
        clean_line = line.lower().translate(str.maketrans('', '', string.punctuation))
        words = clean_line.split()
        
        # Count matched positive and negative words in the review
        pos_count = sum(1 for w in words if w in pos_words)
        neg_count = sum(1 for w in words if w in neg_words)
        
        # 3. Classify the review based on the question rules
        if pos_count > neg_count:
            pos_reviews += 1
        elif pos_count < neg_count:
            neg_reviews += 1
        else:
            neu_reviews += 1

# 4. Print the results to answer Questions 2, 3, and 4
print(f"Question 2 - Positive reviews: {pos_reviews}")
print(f"Question 3 - Negative reviews: {neg_reviews}")
print(f"Question 4 - Neutral reviews:  {neu_reviews}")
```

```python
import string
from collections import Counter

# Words to count frequency
q5_words = ['like', 'great', 'good', 'wonderful']
q6_words = ['bad', 'hard', 'plot', 'slow']

word_counts = Counter()

# Read the reviews file and count word occurrences
with open('amazon_reviews', 'r', encoding='utf-8') as f:
    for line in f:
        # Lowercase and remove punctuation (adjust if your course uses simple .split() instead)
        clean_line = line.lower().translate(str.maketrans('', '', string.punctuation))
        words = clean_line.split()
        word_counts.update(words)


print("Question 5 - Positive word counts:")
for w in q5_words:
    print(f"  {w}: {word_counts[w]}")
least_q5 = min(q5_words, key=lambda w: word_counts[w])
print(f"--> Least matched positive word: {least_q5}\n")


print("Question 6 - Negative word counts:")
for w in q6_words:
    print(f"  {w}: {word_counts[w]}")
least_q6 = min(q6_words, key=lambda w: word_counts[w])
print(f"--> Least matched negative word: {least_q6}")
```
