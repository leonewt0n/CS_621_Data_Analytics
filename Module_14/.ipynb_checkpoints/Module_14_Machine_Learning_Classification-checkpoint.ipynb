```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix

# Load training and test datasets
train_df = pd.read_csv('training_digits.csv')
test_df = pd.read_csv('test_digits.csv')

X_train = train_df.drop('label', axis=1)
y_train = train_df['label']
X_test = test_df.drop('label', axis=1)
y_test = test_df['label']

# Train a classifier (e.g., Random Forest, KNN, Decision Tree, or Logistic Regression)
model = RandomForestClassifier(random_state=0)
model.fit(X_train, y_train)

# Predict on the test data
y_pred = model.predict(X_test)

# Compute per-digit accuracy (recall per class)
cm = confusion_matrix(y_test, y_pred)
per_digit_accuracy = pd.Series(cm.diagonal() / cm.sum(axis=1), index=range(10))
print(per_digit_accuracy)

# Verify the multiple-choice statements
print("\nAccuracy of digit 1 > digit 9:", per_digit_accuracy[1] > per_digit_accuracy[9])
print("Accuracy of digit 9 > digit 1:", per_digit_accuracy[9] > per_digit_accuracy[1])
print("Accuracy of digit 3 > digit 0:", per_digit_accuracy[3] > per_digit_accuracy[0])
print("Accuracy of digit 3 > digit 7:", per_digit_accuracy[3] > per_digit_accuracy[7])
```
