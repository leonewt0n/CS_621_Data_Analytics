```python
# The input text; Feel free to supply your own text. 
text = """
to be or not to be
that is the question
to be or to code
"""


```

```python
# The mapper creates many (word, 1) tuples.
def mapper(text):
    results = []

    for line in text.strip().split("\n"):
        for word in line.lower().split():
            results.append((word, 1))

    return results
```

```python
mapped = mapper(text)
```

```python
type(mapped)
```

```python
# Show every tuple created.
for item in mapped:
    print(item)
```

```python
# The shuffle function creates several key:value pairs; Here the key is the word and its value is a list of 1s.
def shuffle(mapped_data):
    grouped = {}

    for key, value in mapped_data:
        if key not in grouped.keys():
            grouped[key] = []

        grouped[key].append(value)

    return grouped
```

```python
grouped = shuffle(mapped)
```

```python
type(grouped)
```

```python
grouped
```

```python
# The reducer function aggregates all the 1s as the sum.
def reducer(grouped_data):
    results = {}

    for key, values in grouped_data.items():
        results[key] = sum(values)

    return results
```

```python
counts = reducer(grouped)

for word, count in sorted(counts.items()):
    print(f"{word:10} {count}")
```

```python
with open("cs621_module_11_newton.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
# Running MapReduce on Module 11 cleaned reviews
from collections import defaultdict
from functools import reduce

# 1. MAP PHASE: Emit (word, 1) for each word
def mapper(line):
    for word in line.split():
        yield (word, 1)

mapped_pairs = list(mapper(text))

# 2. SHUFFLE / GROUP PHASE: Group values by key
grouped = defaultdict(list)
for key, value in mapped_pairs:
    grouped[key].append(value)

# 3. REDUCE PHASE: Sum the counts for each word
def reducer(accumulated, item):
    key, values = item
    accumulated[key] = sum(values)
    return accumulated

word_counts = reduce(reducer, grouped.items(), {})
```

```python
print(word_counts)
```
