```python
import matplotlib.pyplot as plot
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Given a list of numerical values, compute their mean value.
def my_mean(m):
    total = 0.0
    for x in m:
        total = total + float(x)
    return float(total)/len(m)
```

```python
# Given X and Y of numerical values, compute the w1 coefficient of the linear regression model.
def my_w1(X, Y):
    x_mean = my_mean(X)
    y_mean = my_mean(Y)
    num = 0.0
    den = 0.0
    for i in  range(len(X)):
        num = num + (X[i]- x_mean) * (Y[i] - y_mean)
        den = den + (X[i]- x_mean)*(X[i]- x_mean)
    return num/den
```

```python
# Given X and Y of numerical values, compute the w0 coefficient of the linear regression model. 
def my_w0(X, Y):
    mean_y = my_mean(Y)
    mean_x = my_mean(X)
    return mean_y - my_w1(X, Y)*mean_x
```

```python
# Create S and T, two list of integers.
S = [3, 8, 9, 13, 3, 6, 11, 21, 1, 16]
T = [30, 57, 64, 72, 36, 43, 59, 90, 20, 83]
```

```python
# Pair up S[i] and T[i] as a point. Plot all paired points. 
plot.axis([0, 25, 0, 100])
plot.plot(S, T, 'o')
```

```python
# Given S and T, compute the w1 and w0 of the regression model. 
w1 = my_w1(S, T)
w0 = my_w0(S, T)
```

```python
# Predict T values using S only; 
w1 = my_w1(S, T)
w0 = my_w0(S, T)
t_predicted= []
for x in S:
    y = w1 * x + w0
    t_predicted.append(y)
```

```python
# Contruct the regression line of 1000 points.
x_samples = np.linspace(0, 25, 1000)
y_samples = []
for x in x_samples:
    y_samples.append(w0+w1*x)
```

```python
# Plot the input data and the regression line.
plot.axis([0, 25, 0, 100])
plot.plot(S, T, 'o')
plot.plot(x_samples, y_samples, color='green')
```

```python
# Each red segment represents an error (difference between true t and predicted t)
plot.plot(S, T, 'o')
plot.plot(x_samples, y_samples, color = 'green')
for i in range(len(S)):
    x_pair = [S[i], S[i]]
    y_pair = [T[i], t_predicted[i]]
    plot.plot(x_pair, y_pair, color = 'red')
```

```python
# Create a prediction function with given w1, w0, and X[] values. 
def my_prediction(w1, w0, X):
    result = []
    for x in X:
        result.append(w1*x + w0)
    return result
```

```python
# Predict Y values of S. 
T_predicted = my_prediction(w1, w0, S)
```

```python
# Compute the mean absolute error.
import math
def abs_error(Y, Y_head):
    error = 0
    for i in range(len(Y)):
        error = error + math.fabs(Y[i] - Y_head[i])
    return error/len(Y)
```

```python
print(abs_error(T, T_predicted))
```

```python
# Compute the mean squared error. 
import math
def squared_error(Y, Y_head):
    error = 0
    for i in range(len(Y)):
        error = error + math.pow((Y[i] - Y_head[i]), 2)
    return error/len(Y)
```

```python
print(squared_error(T, T_predicted))
```

```python
# Take the squared root of the mean squared error. 
import math
def square_root_error(Y, Y_head):
    error = 0
    for i in range(len(Y)):
        error = error + math.pow((Y[i] - Y_head[i]), 2)
    return math.sqrt(error/len(Y))
```

```python
print(square_root_error(T, T_predicted))
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# 1. Load and clean the data
df = pd.read_csv('flights_1000.csv')
df_clean = df.dropna(subset=['DepDelay', 'ArrDelay'])

# 2. Define features (X) and target (y)
X = df_clean[['DepDelay']]
y = df_clean['ArrDelay']

# 3. Train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# 4. Calculate parameters and error
w1 = model.coef_[0]
w0 = model.intercept_
y_pred = model.predict(X)
mae = mean_absolute_error(y, y_pred)

print(f"w1 (coefficient): {w1:.4f}")
print(f"w0 (intercept): {w0:.4f}")
print(f"MAE: {mae:.4f}")
```
